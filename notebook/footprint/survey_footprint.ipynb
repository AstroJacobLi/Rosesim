## Show the footprint of Core Community Surveys of Roman

```python
import numpy as np
import healpix as hp
import healsparse as hsp
data = hsp.HealSparseMap.read('../../data/map_f158_all.hsp')
```

```python
from matplotlib.lines import Line2D
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
from astropy.coordinates import SkyCoord
import astropy.units as u
from dustmaps.sfd import SFDQuery
import skyproj
from matplotlib.colors import ListedColormap, BoundaryNorm
```

```python
# --- Figure & projection ---
fig, ax = plt.subplots(figsize=(12, 10))
plt.subplots_adjust(top=0.90, right=0.85)

sp = skyproj.MollweideSkyproj(ax=ax, lon_0=80, longitude_ticks='positive', n_grid_lat=6)
sp.draw_milky_way(width=15)  # thick equator + dashed ±10° by default

# Custom colormap for valid regions (light blue, alpha=0.2)
cmap = ListedColormap(['deepskyblue'])
sp.draw_hspmap(data,# cmap=cmap, 
               alpha=0.2, 
               xsize=500,
               valid_mask=True, zoom=False)

# To extract boundary, one can use hsp.HealSparseMap.valid_pixels_pos() -> pixel to geometry
# or if skyproj supports it. For now, visualizing the mask is sufficient.
```
